## FizzBuzz

````python
def fizz_buzz_1(n: int) -> None:
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)


def fizz_buzz_2(n: int) -> None:
    for i in range(1, n + 1):
        out = ""
        if i % 3 == 0:
            out += "Fizz"
        if i % 5 == 0:
            out += "Buzz"
        if out == "":
            out = i
        print(out)


n = 20
# fizz_buzz_1(n=n)
fizz_buzz_2(n=n)
````

## Vectors

Neural networks do lots of (simple) linear algebra. As a start, we implement the dot product between two "vectors" (here simply lists).

````python
def my_dotproduct(a, b):
    n = min(len(a), len(b))
    result = 0
    for i in range(n):
        result += a[i] * b[i]
    return result


# Generator expression saves memory over list comprehension
def dotproduct(a, b):
    return sum((ai * bi for ai, bi in zip(a, b)))


vec1 = [1, 0, 0]
vec2 = [0, 1, 0]

print(my_dotproduct(vec1, vec2))
print(dotproduct(vec1, vec2))
````

## Matrices

We implement a class that has functionality to do matrix multiplication and matrix addition.

Quick reminder: lists in Python are passed by reference! (mutable)

````python
import copy

a = [1, 2, 3]
b = a
b[0] = 42
print(a, b)

# Same code with deep copy!
a = [1, 2, 3]
b = copy.deepcopy(a)
# b = [x for x in a]
b[0] = 42
print(a, b)
````

````python
class Matrix:
    def __init__(self, vals: list[list[int]]) -> None:
        # Copy pitfall! Deep copy required because of pass/call-by-reference: mutable objects can be modified from outside (see code block above!)
        self.vals = copy.deepcopy(vals)

    def matmul(self, other):
        # self.vals: m x n matrix
        # other.vals: n x o matrix
        m = len(self.vals)
        n = len(other.vals)
        o = len(other.vals[0])
        result = [[0] * o for _ in range(m)]  # m x o result dim
        # print(result)  # initialized with zeros
        A = self.vals
        B = other.vals

        for i in range(m):
            for j in range(o):
                result[i][j] = sum(A[i][k] * B[k][j] for k in range(n))
        return Matrix(result)

    def __repr__(self) -> str:
        return str(self.vals)

    def __add__(self, other):
        return Matrix(
            [
                [aij + bij for aij, bij in zip(ai, bi)]
                for ai, bi in zip(self.vals, other.vals)
            ]
        )


vals = [[1, 2], [3, 4]]
mat1 = Matrix(vals)
print(mat1.vals)

vals[0][0] = 1000
print(mat1.vals, "\n")  # no effect due to deep copy!

mat2 = Matrix([[2, 3], [4, 5]])
print(f"mat1 = {mat1}")
print(f"mat2 = {mat2}")
print("Add:", mat1 + mat2)
print("Matmul:", mat1.matmul(mat2))
````
